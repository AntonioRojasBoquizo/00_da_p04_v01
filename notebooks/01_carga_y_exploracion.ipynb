## Problemas detectados en la inspección inicial

### CSV (`bank-additional.csv`): 43.000 filas, 24 columnas

**Estructura y columnas**
- Columna extra `Unnamed: 0` no descrita en el enunciado. No coincide con el índice y tiene 1.968 valores repetidos, por lo que no sirve como identificador. **Decisión:** eliminar.
- Nombres de columnas con puntos (`emp.var.rate`, `cons.price.idx`, `cons.conf.idx`, `nr.employed`) e `id_` con guion bajo final. **Decisión:** renombrar a formato limpio.

**Tipos de datos**
- `cons.price.idx`, `cons.conf.idx`, `euribor3m` y `nr.employed` están como texto porque usan **coma decimal** (ej. `93,994`). **Decisión:** reemplazar `,` por `.` y convertir a número.
- `date` está como texto y **en español** (ej. `2-agosto-2019`). **Decisión:** traducir los meses con un diccionario y convertir con `to_datetime`.
- `age`, `default`, `housing` y `loan` son `float64` por los nulos. **Decisión:** corregir tras tratar los nulos.

**Nulos**
| Columna | Nulos | % aprox. |
|---|---|---|
| euribor3m | 9.256 | 21,5 % |
| default | 8.981 | 20,9 % |
| age | 5.120 | 11,9 % |
| education | 1.807 | 4,2 % |
| housing | 1.026 | 2,4 % |
| loan | 1.026 | 2,4 % |
| cons.price.idx | 471 | 1,1 % |
| job | 345 | 0,8 % |
| date | 248 | 0,6 % |
| marital | 85 | 0,2 % |

- `housing` y `loan` tienen nulos **en las mismas filas** (comprobado).

**Duplicados**
- 0 duplicados completos (también sin contar `Unnamed: 0`).
- 0 duplicados por `id_`.

**Valores especiales y atípicos**
- `pdays = 999` en más del 75 % de las filas (significa "no contactado antes"). **Decisión:** crear `contactado_antes` y tratar 999 como nulo.
- `default`: 34.016 ceros, solo **3 unos** y 8.981 nulos. Sin variabilidad útil. **Decisión:** excluir del análisis y justificarlo en el informe.
- `duration = 0` en 4 filas. **Decisión:** conservar y anotar.
- `campaign` con máximo 56 (cola larga). **Decisión:** revisar outliers en la Sesión 4.
- `age` mínimo 17 y máximo 98; 10 casos con más de 90 años. **Decisión:** conservar (valores posibles).
- `age` tiene 11,9 % de nulos. **Decisión:** definir tratamiento en la Sesión 4.

**Inconsistencias en categorías**
- `marital` y `poutcome` están en MAYÚSCULAS; el resto de columnas de texto, en minúsculas. **Decisión:** unificar a minúsculas.
- `education` incluye `illiterate` con solo 18 casos. **Decisión:** conservar o agrupar (a decidir).
- No se detectan valores tipo `unknown`, `?` o espacios en blanco; los vacíos son `NaN`.

**Variable objetivo**
- `y` solo toma `yes` / `no`, pero está **desbalanceada**: 4.844 `yes` (≈11,3 %) frente a 38.156 `no`. Se mencionará en el informe.
- `duration` solo se conoce después de la llamada: sirve para describir, no para predecir.

**Latitud y longitud**
- Rangos: latitud 24,4 a 49,4 y longitud -125,0 a -66,9, que corresponden a **Estados Unidos**, no a Portugal. Incoherente con el enunciado. **Decisión:** no usar como análisis geográfico de Portugal y anotar la incoherencia en el informe.

### Excel (`customer-details.xlsx`): 3 hojas

| Hoja | Filas | Columnas | Nulos | Rango de `Dt_Customer` |
|---|---|---|---|---|
| 2012 | 20.115 | 7 | 0 | 2012-01-01 → 2012-12-31 |
| 2013 | 8.965 | 7 | 0 | 2013-01-01 → 2013-12-31 |
| 2014 | 14.090 | 7 | 0 | 2014-01-01 → 2014-12-31 |
| **Total** | **43.170** | | | |

- Las 3 hojas tienen **las mismas columnas**, por lo que se pueden concatenar.
- También incluyen una columna `Unnamed: 0`. **Decisión:** eliminar.
- `Dt_Customer` ya viene como fecha y cada hoja coincide con su año.
- `Income` entre 5.852 y 180.791, sin nulos ni atípicos claros (revisado en la hoja 2012).
- `Kidhome` y `Teenhome` toman valores 0, 1 y 2.
- `NumWebVisitsMonth` hasta 32, con distribución casi uniforme (posible dato sintético; no sobreinterpretar).
- 0 `ID` duplicados entre las 3 hojas.

### Compatibilidad entre datasets
- IDs en común: **43.000**
- IDs solo en el CSV: **0**
- IDs solo en el Excel: **170**
- **Decisión:** unión `left` desde el CSV. Resultado esperado: 43.000 filas.
- Las fechas del Excel (2012-2014) son anteriores a las del CSV (2015-2019), lo que permite calcular la **antigüedad del cliente**.

### Resumen de decisiones para la Sesión 3
1. Eliminar `Unnamed: 0` (CSV y Excel).
2. Concatenar las 3 hojas del Excel.
3. Renombrar columnas a formato limpio.
4. Convertir a número las 4 columnas con coma decimal.
5. Traducir y convertir `date` a fecha.
6. Pasar `marital` y `poutcome` a minúsculas.
7. Excluir `default` del análisis.
8. Tratar `pdays = 999` y crear `contactado_antes`.
9. Unir con `left` y validar que quedan 43.000 filas.
